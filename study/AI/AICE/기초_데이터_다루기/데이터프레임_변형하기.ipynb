### 그룹화하기
groupby 메소드는 데이터프레임의 같은 값을 하나로 묶어서 통계나 집계 결과를 확안하기 위해 사용한다.
예를 들어 기상 데이터를 날짜별, 도시별로 값을 묶어서 통계나 집계 결과를 확인할 수 있다.

- Split: 그룹별로 데이터를 나눈다.
- Apply: 각 그룹별로 집계 함수를 적용한다.
- Combine: 그룹별 집계 결과를 하나로 합치는 과정이다.

```python
import numpy as np
import pandas as pd

flight = pd.read_csv('Clean_Dataset.csv')
```

```python
# aireline 칼럼 기준으로 그룹화하기
airline_group = flight.groupby('airline')
airline_group
```

그룹화를 수행하면 airline 기준으로 데이터들이 모일 것이라고 생각하지만, 수행한 결과는 데이터프레임으로 그룹까지 생성한 것이다.
따라서 그룹화된 결과를 확인하려면 `groups` 속성을 이용한다.

```python
# 수행 결과 확인하기
airline_group.groups
```

데이터는 그룹화된 것을 확인할 수 있지만 가독성은 좋지 않다.

groupby 내부 메소드를 활용하여 데이터를 확인할 수 있다.
- count : 데이터 개수
- size : 집단별 크기
- sum : 데이터의 합
- mean, std, var : 평균, 표준편차, 분산
min, max : 최솟값, 최댓값

```python
# 데이터 수 확인하기
airline_group.count()
```

데이터프레임의 복수 개의 칼럼을 기준으로 groupby를 활용하면 다중 인덱싱(multi-indexing)을 설정할 수 있다.

```python
# groupby를 활용하여 다중 인덱싱(multi-indexing) 설정하기
flight.groupby(['airline', 'arrival_time']).mean(numeric_only=True)
```

여러 개의 칼럼을 기준으로 groupby한 경우 loc을 이용하여 원하는 인덱스의 데이터만 가져올 수 있다.

```python
# groupby 후 원하는 데이터만 가져오기
flight.groupby(['airline', 'arrival_time']).mean(numeric_only=True).loc[[('AirAsia', 'Evening')]]
```

### 인덱스로 그룹화하기
인덱스를 설정하여 그룹화할 때도 groupby 메소드의 레벨(level) 설정이 가능하다.
- set_index : 칼럼을 인덱스로 변경하는 경우에 사용
- reset_index : 인덱스를 초기화

```python
# set_index로 인덱스 설정하기
flight.set_index(['airline', 'arrival_time'])
```

### Aggregate로 집계하기
aggregate 메소드를 활용하면 데이터프레임의 값을 다양하게 집계하여 한 번에 볼 수 있다.

```python
# aggregate 메소드 이용하여 groupby 후 평균값과 최댓값 확인하기
flight.set_index(['airline', 'arrival_time']).groupby(level=[0,1]).aggregate([np.min, np.max])
```

### 피벗 테이블 생성하기
'(축을 중심으로) 회전한다'는 사전적 의미처럼 pivot과 pivot_table 메소드는 행 데이터를 열 데이터로 회전할 수 있다.

```python
# pivot/pivot_table을 위한 데이터프레임 만들기
pivot_data = pd.DataFrame({'cust_id': ['cust_1', 'cust_1', 'cust_1', 'cust_2', 'cust_2', 'cust_2', 'cust_3', 'cust_3', 'cust_3'],
                           'prod_cd': ['p1', 'p2', 'p3', 'p1', 'p2', 'p3', 'p1', 'p2', 'p3'],
                           'grade': ['A', 'A', 'A', 'A', 'A', 'A', 'B', 'B', 'B'],
                           'purch_amt': [30, 10, 0, 40, 15, 30, 0, 0, 10]})
pivot_data
```

pivot을 사용하여 해당 데이터프레임을 변경해보자.
pivot은 기본적으로 index, columns, values를 지정하여 구성할 수 있다.

```python
# pivot 활용하기
pivot_data.pivot(index = 'cust_id', columns='prod_cd', values='purch_amt')
```

pivot과 pivot_table의 기능은 동일하지만, pivot_table은 pivot과 다르게 aggfunc를 지정할 수 있다.
aggfunc을 지정할 때 중복값이 존재하는 경우 pivot은 사용할 수 없고 pivot_table만 사용할 수 있다.

```python
try:
    pivot_data.pivot(index='grade', columns='prod_cd', values='purch_amt')
except Exception as e:
    print(f"Error occurred: {e}")
```

pivot을 수행하면 위와 같이 인덱스에 중복이 있어 변경할 수 없다는 에러 메시지가 나타난다.
실제 인덱스로 쓰인 grade는 중복값을 가지고 있다.
하지만 pivot_table을 동일하게 수행할 경우 평균값이 결과로 보여진다.

```python
# pivot_table 활용하기
pivot_data.pivot_table(index='grade', columns='prod_cd', values='purch_amt')
```

pivot_table에서 aggfunc의 기본은 평균값이고, 따로 지정할 수도 있다.

아래는 aggfunc을 합계로 지정한 예시이다.

```python
# pivot_table의 aggfunc 지정하기
pivot_data.pivot_table(index='grade', columns='prod_cd', values='purch_amt', aggfunc=np.sum)
```

### 인덱스 및 칼럼 레벨 변경하기
데이터프레임을 변형하는 또 다른 방법으로 stack/unstack 메소드를 알아보자.
'쌓다'라는 의미로 stack은 칼럼 레벨에서 인덱스 레벨로 데이터프레임을 변경한다.
데이터를 행의 레벨로 쌓아 올리는 개념으로 이해하면 된다.

반대로 unstack은 인덱스 레벨에서 칼럼 레벨로 위치를 변경하여 데이터를 쌓아 올린다.

```python
# stack/unstack을 위한 데이터프레임 만들기
stack_data = pd.DataFrame({
    'Location': ['Seoul', 'Seoul', 'Seoul', 'Kyounggi', 'Kyounggi', 'Busan', 'Seoul', 'Seoul', 'Busan', 'Kyounggi', 'Kyounggi', 'Kyounggi'],
    'Day': ['Mon', 'Tue', 'Wed', 'Mon', 'Tue', 'Mon', 'Thu', 'Fri', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Ranfall': [100, 80, 1000, 200, 200, 100, 50, 100, 200, 100, 50, 100],
    'Rainfall_precipitation': [80, 70, 90, 10, 20, 30, 50, 90, 20, 80, 50, 10],
    'Temp': [32, 27, 32, 31, 30, 28, 27, 25, 26, 33, 34, 31]
    })
stack_data
```

Location과 Day를 인덱스로 설정하고 새로운 데이터프레임을 만든다.

```python
# 인덱스 설정하기
new_stack_data = stack_data.set_index(['Location', 'Day'])
new_stack_data
```

unstack은 인덱스 레벨에서 칼럼 레벨로 위치를 변경하여 데이터를 쌓아 올리는 것이다.

```python
# unstack 확인하기
new_stack_data.unstack(0)
```

첫 번째 인덱스 Location이 칼럼 레벨로 올라간 것을 확인할 수 있다.
인덱스에도 레벨(Depth)이 있어서 `unstack(1)`을 수행해 보면 두 번째 인덱스였던 요일이 칼럼 레벨로 올라간다.
