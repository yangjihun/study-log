### 칼럼명으로 데이터 선택하기
데이터의 칼럼이 많은 경우, 특정 칼럼을 기준으로 데이터프레임의 데이터를 추출할 수 있다.

```python
import numpy as np
import pandas as pd

flight = pd.read_csv('Clean_Dataset.csv', encoding='cp949')
```

```python
# 하나의 칼럼만 선택하기
flight[['departure_time']]
```

```python
# 여러 개의 칼럼 선택하기
flight[['airline', 'departure_time', 'source_city']]
```

### 행 범위를 지정하여 데이터 선택하기
데이터프레임에서 슬라이싱(Slicing)을 이용하여 행(row) 범위의 데이터를 선택할 수 있다.

```python
# 슬라이싱을 이용하여 10행부터 20행까지 데이터 가져오기
flight[10:21]
```

### 특정 행, 열의 범위를 선택하여 데이터를 선택하기
데이터프레임에서 특정 행과 열의 범위를 지정하여 데이터를 선택하는 방법으로 `loc`과 `iloc`이 있다.

| loc(location) | iloc(integer location) |
|-|-|
| 데이터프레임의 행이나 열에 레이블로 접근 | 데이터프레임의 행이나 열에 인덱스 값으로 접근 |
| 인덱스 및 칼럼명을 통해 지정하는 방법 | 인덱스를 활용해 지정하는 방법 |
| 설정한 인덱스를 그대로 사용 | 0 based index로 사용 |

```python
# 인덱스 새롭게 지정하기
flight.index = np.arange(100, 300253)
flight
```

```python
# loc 사용하기
flight.loc[[102, 202, 302]]
```

```python
# iloc 사용하기
flight.iloc[[102, 202, 302]]
```

### 조건으로 데이터 선택하기
Boolean 연산을 통해 원하는 데이터만 추출할 수도 있다.

```python
# boolean 연산으로 조건을 만족하는 데이터만 추출하기
flight_extract = flight[(flight['price']>12000) & (flight['airline']=='Air_India')]
flight_extract.head()
```

조건을 자주 변경해야 하거나 조건이 너무 많으면 조건을 변수로 저장하여 사용할 수도 있다.

```python
# 조건을 별도로 지정하여 데이터 추출하기
# 가격 : 12000 초과
price_Tag = flight['price']>12000

# 항공사 이름 : Air_India
airline_tag = flight['airline']=='Air_India'
flight_extract2 = flight[price_Tag & airline_tag].head()
flight_extract2
```
