### 데이터 살펴보기
데이터프레임 형태로 만들어진 전체 데이터를 먼저 살펴보자.

```python
# 넘파이(Numpy), 판다스(Pandas) 불러오기
import pandas as pd
import numpy as np

# read_csv 함수로 "Clean_Dataset.csv" 불러오기
flight = pd.read_csv('./Clean_Dataset.csv', encoding = "cp949")
flight
```

각 항목이 의미하는 내용은 다음과 같다.
1. 칼럼(Column): 열 레이블에 대한 정보를 보유한다.
2. 인덱스(Index): 행(Row) 인덱스에 대한 정보를 보유한다.
3. 생성된 데이터프레임의 행과 열의 수를 표시한다.

데이터프레임 이름만 실행할 경우 전체 데이터가 출력되지만, `head`, `tail` 메소드를 활용하면 데이터 일부만 확인할 수도 있다.

```python
# head 확인하기
flight.head()
```

```python
# tail 확인하기
flight.tail()
```

`head`와 `tail` 값을 별도로 지정하지 않으면 5줄을 보여주고, 값을 지정하면 지정한 행의 수만큼 보여준다.

```python
# head로 상위 10개의 데이터 확인하기
flight.head(n=10)
```

### 데이터프레임의 기본 정보 확인하기
판다스는 데이터 분석을 위한 라이브러리로 기본적인 통계 정보를 제공한다.

`shape`의 속성 변환값은 튜플로 존재하며, 행(row)과 열(column)의 개수를 튜플 형태로 변환해준다.

```python
# shape 확인하기
flight.shape
```

`columns`는 데이터프레임을 구성하는 칼럼명을 확인할 때 사용한다.

```python
# columns 확인하기
flight.columns
```

`info`는 데이터 유형, 각 칼럼의 데이터 수 등을 한 번에 볼 수 있어 데이터의 구성이나 결측치 유무를 확인할 수 있다.

```python
# info 확인하기
flight.info()
```

`describe`는 데이터 칼럼별 요약 통계량을 나타낸다. float나 int처럼 숫자형 데이터의 통계치를 확인하는 데 유용하다.(count: 데이터 개수, mean: 평균값, std: 표준편차, min: 최솟값, max: 최댓값, 4분위값)

```python
# describe 확인하기
flight.describe()
```

`info` 함수에서 확인한 데이터 유형 중 float와 int형인 4개의 칼럼만 describe에서 값이 나온 것을 확인할 수 있다.
 
범주형 데이터 정보를 확인하려면 `all`로 지정한다.

`flight.describe(indlude='all)`

`dtype`는 데이터프레임이 가지는 데이터 형태의 종류를 나타낸다.

```python
flight.dtypes
```

`value_counts`는 지정된 칼럼의 값(value)에 대한 발생 횟수를 나타낸다.

```python
# value_counts 확인하기
flight['source_city'].value_counts()
```
